### Introduction

It is a feature extraction technique and its main goal is to reduce Curse of Dimensionality it reduces the Dimension. PCA tries to reduce a higher dimensional data into lower dimensional data it finds the best possible dimension to get the best possible columns or data. It is a technique which can transform higer dimensional data into lower dimesional data while preserving the essence of the data.
- Benifits of Using PCA
  1. Higher dimensional dataset into lower dataset size is reduced so automatically the algorithm will run faster becuase size is reduced.
  2. Visiualization (it helps to reduce the data in to different dimensional and then plot the graph)
