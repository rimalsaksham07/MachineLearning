Curse of Dimensionality

The curse of Dimensionality is one of the important concept of Machine Learning  where if the optimal soltion of and given dataset is with 4 column but when we are reducing the column we reduced it into 3 column then the model will not work as it should be so we wont get the desired accuracy this is curse of dimensionality also if we take more column than we are required for optimal solution then also it is an curse of dimensionality.
